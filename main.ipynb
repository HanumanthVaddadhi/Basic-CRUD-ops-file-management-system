```python
from pathlib import Path
import os
#File Creation
def create_file():
    try:
        file_name = input("Enter the name of the file you want to create : ")
        file_path = Path(file_name)
        if file_path.exists():
            print(f"File {file_name} already exists.")
        else:
            file_path.touch()
            print(f"File {file_name} created successfully.")
    except Exception as err:
        print(f"An error occured : {err}")

#File Reading
def read_file():
    try:
        file_name = input("Enter the name of the file you want to read : ")
        file_path = Path(file_name)
        if file_path.exists():
            with open(file_path, 'r') as file:
                print(file.read())
        else:
            print(f"File {file_name} does not exist.")
    except Exception as err:
        print(f"An error occured : {err}")

#File Updating
def update_file():
    try:
        file_name = input("Enter the name of the file you want to update : ")
        file_path = Path(file_name)
        if file_path.exists():
            with open(file_path, 'w') as file:
                file.write(input("Enter the content to update : "))
                print(f"File {file_name} updated successfully.")
        else:
            print(f"File {file_name} does not exist.")
    except Exception as err:
        print(f"An error occured : {err}")

#File Deleting
def delete_file():
    try:
        file_name = input("Enter the name of the file you want to delete: ")
        file_path = Path(file_name)
        if file_path.exists():
            os.remove(file_path)
            print(f"File {file_name} deleted successfully.")
        else:
            print(f"File {file_name} does not exist.")
    except Exception as err:
        print(f"An error occured : {err}")
```

```python
#Main Screen(UI/UX)
print("Enter 1 for file creation, 2 for file reading, 3 for file updating, 4 for file deleting")
n=int(input("Enter your choice"))
if n==1:
    create_file()
elif n==2:
    read_file()
elif n==3:
    update_file()
elif n==4:
    delete_file()
else:
    print("Invalid choice.")
```
